```python
def odd_even(x):
    if x%2==0:
        print(x, " is even")
    else:
        print(x, "is odd")
```

```python
odd_even(3)
```

```python
odd_even(0)
```

```python
odd_even(2+2)
```

```python
# Lambda function with filter
```

```python
l1 = [87,56,90,34,2,5,7,1,7,98]
final_list = list(filter(lambda x: (x%2!=0),l1))
```

```python
final_list
```

```python
final_list_even = list(filter(lambda x: (x%2==0),l1))
```

```python
final_list_even
```

```python
#Lambda with map
```

```python
l1 = [1,2,3,4,5,6,7,8]
```

```python
l2 = list(map(lambda x: x*2,l1))
```

```python
l2
```

```python
# Lambda with reduce
```

```python
from functools import reduce
```

```python
sum = reduce(lambda x,y: x+y,l1)
```

```python
sum
```
