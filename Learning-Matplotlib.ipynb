```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.arange(1,11)
x
```

```python
y = x*2
```

```python
y
```

```python
plt.plot(x,y)
```

```python
plt.show()
```

```python
plt.plot(x,y,color='r',linestyle=':', linewidth=5)
plt.title("linear relation")
plt.xlabel("x-label")
plt.ylabel("y-label")
```

```python
y1 = 2*x
y2 = 3*x
```

```python
plt.plot(x,y1,color='r',linestyle=':')
plt.plot(x,y2,color='y', linewidth = 4)
plt.title('Two lines in a plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid(True)
plt.show()
```

```python
#adding two sub-plots
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.subplot(1,2,2)
plt.plot(x,y2)
plt.show()
```

```python
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.subplot(2,1,2)
plt.plot(x,y2)
plt.show()
```

```python
student = {'bob': 83, 'mat':32,'sam': 43,'julia': 76}
names  = list(student.keys())
values = list(student.values())
```

```python
plt.bar(names,values)
plt.title('Distribution of student marks')
plt.xlabel('Names of students')
plt.ylabel('marks of students')
plt.grid(True)
plt.show()
```

```python
plt.barh(names,values,color = 'g')
plt.title('Distribution of student marks')
plt.xlabel('Names of students')
plt.ylabel('marks of students')
plt.grid(True)
plt.show()
```
