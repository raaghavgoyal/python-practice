```python
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
fmri = sns.load_dataset("fmri")
fmri.head()
```

```python
sns.lineplot(x = "timepoint", y="signal",data=fmri)
plt.show()
```

```python
fmri.shape
```

```python
sns.lineplot(x = "timepoint", y="signal",data=fmri,hue = "event")
plt.show()
```

```python
sns.lineplot(x = "timepoint",y = "signal", data = fmri, hue = "event", style = "event")
plt.show()
```

```python
sns.lineplot(x= "timepoint",y="signal",hue="event", style = "event",markers = True, data= fmri)
plt.show()
```

```python
import pandas as pd
sns.set(style = "whitegrid")
pokemon = pd.read_csv('pokemon.csv')
```

```python
sns.barplot(x = 'is_legendary',y = 'speed', data = pokemon)
plt.show()
```

```python
sns.barplot(x="is_legendary",y="weight_kg",data= pokemon, color ='orange')
plt.show()
```

```python
sns.barplot(x='is_legendary',y='speed',hue='generation',data = pokemon)
plt.show()
```

```python
sns.barplot(x='is_legendary',y="weight_kg",hue = 'generation',data = pokemon, palette = 'Blues_d')
plt.show()
```
