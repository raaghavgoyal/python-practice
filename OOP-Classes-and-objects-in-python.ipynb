```python
class Phone:
    def set_color(self,color):
        self.color = color
    def set_cost(self,cost):
        self.cost = cost
    def show_color(self):
        return self.color
    def show_cost(self):
        return self.cost
    def make_call(self):
        print("I am making a call")
    def play_game(self):
        print("I am playing a game")
```

```python
p2 = Phone()
```

```python
p2.set_color("red")
```

```python
p2.set_cost(5000)
```

```python
p2.show_color()
```

```python
p2.show_cost()
```

```python
p2.make_call()
```

```python
p2.play_game()
```

```python
class Employee:
    def __init__(self, name, age, salary, gender):
        self.name = name
        self.age = age
        self.salary = salary
        self.gender = gender

    def show_employee_details(self):
        print("Name of employee is ", self.name)
        print("Age of employee is ",self.age)
        print("Salary of employee is ",self.salary)
        print("Gender of employee is ",self.gender)
```

```python
e1 = Employee('Ram', 32,50000,'Male')
```

```python
e1
```

```python
e1.show_employee_details()
```
