```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
import datetime as dt
```

```python
#importing main dataset
df = pd.read_csv('covid_19_india.csv', parse_dates = ['Date'], dayfirst = True)
```

```python
df.head()
```

```python
#keeping only required columns
df = df[['Date','State/UnionTerritory','Cured','Deaths','Confirmed']]
#remaining column names
df.columns = ['date','state','cured','deaths','confirmed']
```

```python
df.head()
```

```python
df.tail()
```

```python
today = df[df.date == '2020-07-17']
```

```python
today.head()
```

```python
#sorting data w.r.t. number of confirmed cases
max_confirmed_cases = today.sort_values(by = 'confirmed',ascending = False)
```

```python
max_confirmed_cases
```

```python
top_states_confirmed = max_confirmed_cases[:5]
```

```python
top_states_confirmed
```

```python
sns.set(rc = {'figure.figsize':(7,7)})
sns.barplot(x='state',y='confirmed',data = top_states_confirmed,hue = 'state')
plt.show()
```

```python
top_states_death = today.sort_values(by='deaths',ascending = False)
top_states_death = top_states_death[:5]
top_states_death
```

```python
sns.barplot(x='state',y='deaths',data = top_states_death, hue='state')
plt.show()
```

```python
top_states_cured = today.sort_values(by='cured',ascending = False)
top_states_cured = top_states_cured[:5]
top_states_cured
```

```python
sns.barplot(x = 'state',y = 'cured', data = top_states_cured,hue = 'state')
plt.show()
```
